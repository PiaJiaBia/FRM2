## 按经济状态分组与自相关；pp.122–126

```python
data=pd.DataFrame({'x':[1.,2.,3.,1.,2.,3.],'y':[1.,2.,3.,3.,2.,1.],'regime':['A']*3+['B']*3})
out=regime_correlations(data,['x','y'],'regime')
assert_allclose(out.loc[('A','x'),'y'],1)
assert_allclose(out.loc[('B','x'),'y'],-1)
assert_allclose(autocorrelation(pd.DataFrame({'x':np.arange(10.)}),'x',[1,2]),1)
```
