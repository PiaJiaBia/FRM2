教材 Example 8.2，p.108：两资产10日 VaR。

```python
positions=pd.DataFrame({'amount':[10e6,5e6],'sigma':[.02,.01]})
correlation=pd.DataFrame([[1,.7],[.7,1]])
cov=covariance_from_correlation(positions,'sigma',correlation)
expected=norm.ppf(.99)*np.sqrt(10*(200000**2+50000**2+2*.7*200000*50000))
assert_allclose(portfolio_var(positions,'amount',cov,.99,10),expected)
assert_allclose(joint_default_probability(.02,.03,0),.0006)
print(expected)
```

## 指数隐含相关系数与组合统计量；pp.104–107

```python
data=pd.DataFrame({'weight':[.5,.5],'sigma':[.2,.2]})
assert_allclose(implied_average_correlation(data,'weight','sigma',np.sqrt(.03)),.5)
returns=pd.DataFrame({'a':[-.1,0,.1],'b':[-.2,0,.2]})
moments=portfolio_moments(returns,['a','b'],data,'weight')
assert_allclose(moments['mean'],0)
assert_allclose(moments['volatility'],.15)
```
